```python
def merge(text, next_id, top_pair):
    new_text = []
    i = 0

    while i < len(text) - 1:
        if text[i] == top_pair[0] and text[i + 1] == top_pair[1]:
            new_text.append(next_id)
            i += 2
        else:
            new_text.append(text[i])
            i += 1

    if i < len(text):
        new_text.append(text[i])

    return new_text

def new_count(text):
    count = {}
    for i in range(len(text) - 1):
        pair = (text[i], text[i + 1])
        count[pair] = count.get(pair, 0) + 1
    return count

text = merge(text, 123, top_pair)
```

```python
print( sorted( ((v,k) for k,v in count.items()),reverse=True))
```

```python
next_id = 124
for _ in range(19):
    count = new_count(text)
    if not count:
        break
    top_pair = max(count, key=count.get)
    text = merge(text, next_id, top_pair)
    next_id += 1
```
